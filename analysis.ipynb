```python
import pandas as pd

feature_path = "data/processed/feature/feature_test.parquet"
df = pd.read_parquet(feature_path)

cols = [
    "brand",
    "source_query",
    "brand_styles",
    "product_lines",
    "product_categories",
    "content_type",
    "weighted_engagement_rate",
]
df[cols].sample(20, random_state=0)
```

```python
(
    df.explode("product_categories")
    .groupby(["brand", "product_categories", "content_type"])["weighted_engagement_rate"]
    .agg(["count", "mean"])
    .reset_index()
    .sort_values("count", ascending=False)
)
```
