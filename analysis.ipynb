```python
import pandas as pd

feature_path = "data/processed/feature/feature_test.parquet"
df = pd.read_parquet(feature_path)

cols = [
    "brand",
    "source_query",
    "brand_style",
    "product_line",
    "product_category",
    "content_type",
    "engagement_rate",
]
df[cols].sample(20, random_state=0)
```

```python
df.groupby(["brand", "product_category", "content_type"])["engagement_rate"] \
  .agg(["count", "mean"]) \
  .reset_index() \
  .sort_values("count", ascending=False) 
```
