## Theme 2 · Commerce orientation (locked)

Source: `04b_theme_commerce.ipynb`. Explicit caption commerce cues in a search-visible Nike and Adidas sample (*n*=4,457). Not sales or conversion; not causal.


### Bottom line

Explicit commerce is rare: None 96.5%, Low 3.5%, Medium *n*=4, High 0. Nike shows higher mean intensity than Adidas (0.050 vs 0.022); the gap is context-specific. No clear engagement penalty is established.


### Measure

Commerce Intensity = Purchase CTA + Discovery/traffic CTA + Promo language + Giveaway (0–4), binned None / Low / Medium / High.


### Locked findings

**Prevalence.** Across 4,457 search-visible Nike and Adidas videos: None 96.5%, Low 3.5%, Medium 4 videos, High 0. Component rates: purchase CTA 0.8%, discovery/traffic 0.9%, promo language 1.8%, giveaway 0.1%. Top-performing content here is mostly **non-transactional**, not overt selling.

**Where cues show up**
- Content: `product_promo` stands out (mean intensity ≈ 0.47); then collaboration / showcase / review. Official campaign ≈ 0.
- Clusters: Jordan culture/releases; interactive / pop-culture.
- Style: Retro > Performance > Lifestyle ≈ Technical.
- Product lines: Nike hotspots include Jordan, Air Max, AF1; Adidas still has local highs (e.g. Tracksuit, Superstar, Samba).

**Nike vs Adidas.** Nike shows more explicit commerce orientation in this sample (mean 0.050 vs 0.022; Low share 4.7% vs 2.1%). The gap shows up on purchase / discovery / promo rates and is concentrated in promotional, showcase, collaboration, and some product-line contexts. Do not read as "Nike sells and Adidas does not."

**Commerce vs engagement (descriptive).** None median BRI 1.01 → Low 0.95 → Medium 0.87 (*n*=4). Low looks slightly softer on median BRI than None, but evidence is insufficient for an engagement penalty or dose-response. Top-10% rate is slightly **higher** for Low (11.7%) than None (10.0%).

**Matrix (descriptive)**
- Higher-commerce + higher BRI: collaboration; OOTD / interactive clusters; Air Max, AF1, Tracksuit, Superstar.
- Higher-commerce + lower BRI: `product_promo`, product review; Jordan cluster; Samba / Gazelle.

More promotional content appears more often in the lower-BRI quadrant; some collab/OOTD and product lines sit higher on both axes. Quadrants are descriptive, not causal.


### Use

Treat explicit commerce as a sparse overlay in this sample. Test selling cues in contexts that already coexist with stronger BRI (e.g. collaboration / OOTD-adjacent, selected product lines). Do not infer conversion lift or platform-wide prevalence.
